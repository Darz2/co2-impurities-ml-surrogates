```python
#!/usr/bin/env python

# =============================================================================
#  Usage: PT colormap via semi_emperical_correlations
#  Methods demonstrated:
#    - model.compute_PT_colormap_data()   with method="parachor"
#    - model.compute_PT_colormap_data()   with method="wsd"
#    - model.compute_PT_colormap_data()   with method="rk"
#    - model.plot_PT_colormap()
# =============================================================================

import time

notebook_start = time.perf_counter()
# %pip install -e /home/darshan/A6/PCSAFT_cDFT

import si_units as si
import numpy    as np
import matplotlib.pyplot as plt
import feos
import thermoift.FeosPlugin                  as TC
import thermoift.semi_emperical_correlations as SEC
```

```python
COMPONENT_NAMES = ["carbon dioxide", "hydrogen", "argon"]
FEED_Z          = np.array([0.95, 0.05, 0.00])
 
PARACHOR        = np.array([78.0, 34.0, 57.0])   # CO2, H2, Ar
KIJ_PARACHOR    = 0.0
N_EXP           = 3.87
PHI_WSD         = 1.0    # WSD mixing parameter (1.0 = original WSD)

# Grid resolution inside the two-phase region
N_T = 30
N_P = 30
 
# Temperature range for bubble / dew curves
T_VALS = np.linspace(210, 305, 60)
```

```python
params  = TC.PARAMETERS(COMPONENT_NAMES)
eos     = feos.HelmholtzEnergyFunctional.pcsaft(params)
feed_si = FEED_Z * si.MOL
 
T_bub, P_bub = TC.compute_bubble_curve(eos, T_VALS, feed_si, verbose=True)
T_dew, P_dew = TC.compute_dew_curve(eos, T_VALS, feed_si, verbose=False)
 
Tc_si, Pc_si = TC.compute_CT(feos, eos, feed_si, T_guess=300.0)
Tc_K   = float(Tc_si / si.KELVIN)
Pc_bar = float(Pc_si / si.BAR)
 
print(f"Critical point:  Tc = {Tc_K:.2f} K,  Pc = {Pc_bar:.2f} bar")
```
